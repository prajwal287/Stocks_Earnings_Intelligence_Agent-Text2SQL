# Phase 2B: Download & Load SEC Filing Text (MD&A)

Download actual 10-Q/10-K documents and extract narrative sections.

**Prerequisites: Run PHASE_2A_LOAD_METRICS.ipynb first**

```python
import sys
sys.path.insert(0, '/Users/prajwalchambenandeeshappa/Github_Repos/Stocks_Earnings_Intelligence_Agent-Text2SQL/learning')

import dlt
import duckdb
import logging
import time
from fetch_filing_text import fetch_filing_text, prepare_chunks_for_dlt

logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

print("✅ Imports successful")
```

```python
# Connect to existing DuckDB database (created by Phase 2A)
db_path = '/Users/prajwalchambenandeeshappa/.dlt/pipelines/financial_data_pipeline/sec_filings.duckdb'
conn = duckdb.connect(db_path)

# Verify Phase 2A data exists
try:
    metrics_count = conn.execute("SELECT COUNT(*) FROM sec_filings.financial_metrics").fetchone()[0]
    filings_count = conn.execute("SELECT COUNT(*) FROM sec_filings.sec_filings_metadata").fetchone()[0]
    print(f"✅ Phase 2A data found:")
    print(f"   Metrics: {metrics_count} records")
    print(f"   Filings: {filings_count} records")
except Exception as e:
    print(f"❌ Error: Phase 2A data not found. Run PHASE_2A first!")
    print(f"   {e}")
```

```python
# Get filing URLs from Phase 2A
filings_list = conn.execute("""
    SELECT ticker, accession_number, filing_date, period_end, form, filing_url
    FROM sec_filings.sec_filings_metadata
    ORDER BY filing_date DESC
    LIMIT 3
""").fetchall()

print(f"\n🚀 PHASE 2B: Downloading & Extracting Filing Text")
print(f"\nProcessing {len(filings_list)} filings...\n")

extracted_filings = []
failed = []

for i, (ticker, accession, date, period, form, url) in enumerate(filings_list, 1):
    print(f"[{i}/{len(filings_list)}] {ticker} {form} {date}")
    
    try:
        result = fetch_filing_text(ticker, accession, date, period, form, url)
        if result:
            extracted_filings.append(result)
            print(f"   ✅ {result['chunk_count']} chunks")
        else:
            failed.append((ticker, form))
            print(f"   ⚠️  No MD&A extracted")
    except Exception as e:
        failed.append((ticker, form))
        print(f"   ❌ {str(e)[:50]}")
    
    if i < len(filings_list):
        time.sleep(2)

print(f"\n✅ Extraction: {len(extracted_filings)} successful, {len(failed)} failed")
```

```python
# Prepare chunks
all_chunks = []
for filing in extracted_filings:
    chunks = prepare_chunks_for_dlt(filing)
    all_chunks.extend(chunks)

print(f"📝 Prepared {len(all_chunks)} chunks")

if all_chunks:
    # Load with dlt
    pipeline = dlt.pipeline(
        pipeline_name="financial_data_pipeline",
        destination="duckdb",
        dataset_name="sec_filings"
    )
    
    print("\n📝 Loading chunks into DuckDB...")
    pipeline.run(all_chunks, table_name="filing_text_chunks", write_disposition="replace")
    print(f"✅ {len(all_chunks)} chunks loaded")
else:
    print("❌ No chunks to load")
```

```python
# Verify all tables
conn = duckdb.connect(db_path)

print("\n" + "="*70)
print("PHASE 2 COMPLETE - DATABASE SUMMARY")
print("="*70)

try:
    metrics = conn.execute("SELECT COUNT(*) FROM sec_filings.financial_metrics").fetchone()[0]
    print(f"✅ financial_metrics: {metrics} records")
except:
    print(f"❌ financial_metrics not found")

try:
    filings = conn.execute("SELECT COUNT(*) FROM sec_filings.sec_filings_metadata").fetchone()[0]
    print(f"✅ sec_filings_metadata: {filings} records")
except:
    print(f"❌ sec_filings_metadata not found")

try:
    chunks = conn.execute("SELECT COUNT(*) FROM sec_filings.filing_text_chunks").fetchone()[0]
    print(f"✅ filing_text_chunks: {chunks} records")
except:
    print(f"ℹ️  filing_text_chunks: not created (OK if extractions failed)")

print("\n" + "="*70)
print("✅ PHASE 2 COMPLETE!")
print("Ready for Module 1: Agentic RAG")
print("="*70)

conn.close()
```
