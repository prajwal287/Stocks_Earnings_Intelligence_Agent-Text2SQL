```python
from openai import OpenAI
client = OpenAI()

question = "What was Microsoft's gross margin in fiscal Q2 2026?"

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": question}],
)
print(response.choices[0].message.content)
```
