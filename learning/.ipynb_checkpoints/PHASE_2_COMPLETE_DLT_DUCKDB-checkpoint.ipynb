# Phase 2: Complete Data Loading (XBRL Numbers + Filing Text)

## Single Notebook for Both:
- **Phase 2A:** Load XBRL financial metrics
- **Phase 2B:** Download & load SEC filing text (MD&A)

### Output: 3 DuckDB Tables
1. `financial_metrics` - XBRL numbers (Revenues, Net Income, etc.)
2. `sec_filings_metadata` - Filing dates and URLs
3. `filing_text_chunks` - MD&A text chunks from actual 10-Q/10-K documents

```python
import sys
sys.path.insert(0, '/Users/prajwalchambenandeeshappa/Github_Repos/Stocks_Earnings_Intelligence_Agent-Text2SQL/learning')

import dlt
import duckdb
import logging
import pandas as pd
from typing import List, Dict, Any
from datetime import datetime
import time

from sec_api_orchestrator import FinancialDataPipeline
from fetch_filing_text import fetch_filing_text, prepare_chunks_for_dlt

logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

print("✅ All imports successful")
print(f"   dlt version: {dlt.__version__}")
```

---
## PHASE 2A: Load XBRL Financial Metrics

Extract from SEC API and load structured financial data

```python
TICKERS = ["MSFT", "AAPL", "GOOGL"]
KEY_METRICS = [
    "Revenues",
    "NetIncomeLoss",
    "OperatingIncomeLoss",
    "Assets",
    "Liabilities",
    "StockholdersEquity",
]

print("🚀 PHASE 2A: Loading XBRL Financial Metrics")
print(f"\nFetching from SEC API...\n")

all_metrics_data = []
all_filings_data = []

for ticker in TICKERS:
    print(f"📊 {ticker}...")
    pipeline = FinancialDataPipeline(ticker)
    
    for metric in KEY_METRICS:
        metric_data = pipeline.get_metric(metric, limit=10)
        all_metrics_data.extend(metric_data)
    
    filings_data = pipeline.get_filings(limit=5)
    all_filings_data.extend(filings_data)

print(f"\n✅ Phase 2A: Data fetched")
print(f"   Metrics: {len(all_metrics_data)} records")
print(f"   Filings: {len(all_filings_data)} records")
```

```python
def transform_metric_for_dlt(records: List[Dict]) -> List[Dict]:
    transformed = []
    load_time = datetime.now().isoformat()
    for record in records:
        transformed.append({
            "ticker": record.get("ticker"),
            "concept": record.get("concept"),
            "period_end": record.get("period_end"),
            "filing_date": record.get("filing_date"),
            "form": record.get("form"),
            "value": int(record.get("value", 0)),
            "load_timestamp": load_time,
        })
    return transformed

def transform_filings_for_dlt(records: List[Dict]) -> List[Dict]:
    transformed = []
    load_time = datetime.now().isoformat()
    for record in records:
        transformed.append({
            "ticker": record.get("ticker"),
            "form": record.get("form"),
            "filing_date": record.get("filing_date"),
            "period_end": record.get("period_end"),
            "accession_number": record.get("accession_number"),
            "filing_url": record.get("filing_url"),
            "load_timestamp": load_time,
        })
    return transformed

metrics_transformed = transform_metric_for_dlt(all_metrics_data)
filings_transformed = transform_filings_for_dlt(all_filings_data)

print("✅ Data transformed")
```

```python
pipeline = dlt.pipeline(
    pipeline_name="financial_data_pipeline",
    destination="duckdb",
    dataset_name="sec_filings"
)

print("📝 Loading metrics...")
pipeline.run(metrics_transformed, table_name="financial_metrics", write_disposition="append")
print(f"   ✅ {len(metrics_transformed)} metric records loaded")

print("📝 Loading filing metadata...")
pipeline.run(filings_transformed, table_name="sec_filings_metadata", write_disposition="append")
print(f"   ✅ {len(filings_transformed)} filing records loaded")

print("\n✅ Phase 2A complete!")
```

---
## PHASE 2B: Download & Load SEC Filing Text (MD&A)

Download actual 10-Q/10-K documents and extract narrative sections

```python
db_path = pipeline.default_schema.connection.db_path
conn = duckdb.connect(db_path)

# Get filing URLs from Phase 2A
filings_list = conn.execute("""
    SELECT ticker, accession_number, filing_date, period_end, form, filing_url
    FROM sec_filings.sec_filings_metadata
    ORDER BY filing_date DESC
    LIMIT 3
""").fetchall()

print("🚀 PHASE 2B: Downloading & Extracting Filing Text")
print(f"\nProcessing {len(filings_list)} filings...\n")

extracted_filings = []
failed = []

for i, (ticker, accession, date, period, form, url) in enumerate(filings_list, 1):
    print(f"[{i}/{len(filings_list)}] {ticker} {form} {date}")
    
    try:
        result = fetch_filing_text(ticker, accession, date, period, form, url)
        if result:
            extracted_filings.append(result)
            print(f"   ✅ {result['chunk_count']} chunks")
        else:
            failed.append((ticker, form))
            print(f"   ⚠️  No MD&A extracted")
    except Exception as e:
        failed.append((ticker, form))
        print(f"   ❌ Error: {str(e)[:50]}")
    
    if i < len(filings_list):
        time.sleep(2)  # Rate limiting

print(f"\n✅ Extraction complete: {len(extracted_filings)} successful, {len(failed)} failed")
```

```python
# Prepare chunks for loading
all_chunks = []
for filing in extracted_filings:
    chunks = prepare_chunks_for_dlt(filing)
    all_chunks.extend(chunks)

print(f"📝 Prepared {len(all_chunks)} chunks for loading...\n")

if all_chunks:
    pipeline.run(all_chunks, table_name="filing_text_chunks", write_disposition="append")
    print(f"✅ {len(all_chunks)} chunks loaded into filing_text_chunks table")
else:
    print("⚠️  No chunks to load")
```

## Verify All Data

```python
conn = duckdb.connect(db_path)

print("\n" + "="*70)
print("DATABASE SUMMARY")
print("="*70)

# Check all tables
tables = conn.execute("SELECT table_name FROM information_schema.tables WHERE table_schema = 'sec_filings' AND table_name NOT LIKE 'dlt%'").fetchall()

for table_name, in tables:
    count = conn.execute(f"SELECT COUNT(*) FROM sec_filings.{table_name}").fetchone()[0]
    print(f"✅ {table_name}: {count} records")

# Show metrics summary
print("\n📊 Metrics by company:")
summary = conn.execute("""
    SELECT ticker, COUNT(*) as records, COUNT(DISTINCT concept) as concepts
    FROM sec_filings.financial_metrics
    GROUP BY ticker
    ORDER BY ticker
""").df()
print(summary.to_string(index=False))

# Show chunks summary
try:
    print("\n📄 Filing text chunks:")
    chunks_summary = conn.execute("""
        SELECT ticker, COUNT(*) as chunks, SUM(text_length) as total_chars
        FROM sec_filings.filing_text_chunks
        GROUP BY ticker
    """).df()
    print(chunks_summary.to_string(index=False))
except:
    print("   (No chunks loaded yet)")
```

```python
# Create indexes
print("\n" + "="*70)
print("CREATING INDEXES")
print("="*70)

indexes = [
    ("ticker", "CREATE INDEX IF NOT EXISTS idx_metrics_ticker ON sec_filings.financial_metrics(ticker)"),
    ("concept", "CREATE INDEX IF NOT EXISTS idx_metrics_concept ON sec_filings.financial_metrics(concept)"),
    ("period", "CREATE INDEX IF NOT EXISTS idx_metrics_period ON sec_filings.financial_metrics(period_end)"),
    ("ticker+concept", "CREATE INDEX IF NOT EXISTS idx_metrics_ticker_concept ON sec_filings.financial_metrics(ticker, concept)"),
]

for name, sql in indexes:
    try:
        conn.execute(sql)
        print(f"✅ {name}")
    except:
        pass

# Chunk indexes if table exists
try:
    conn.execute("CREATE INDEX IF NOT EXISTS idx_chunks_ticker ON sec_filings.filing_text_chunks(ticker)")
    print(f"✅ chunks: ticker")
except:
    pass
```

---
## Example Queries

```python
print("\n" + "="*70)
print("EXAMPLE QUERY 1: Revenue Trend")
print("="*70)

result = conn.execute("""
    SELECT ticker, period_end, value/1000000000 as revenue_billions
    FROM sec_filings.financial_metrics
    WHERE concept = 'Revenues'
    ORDER BY ticker, period_end DESC
    LIMIT 6
""").df()

print(result.to_string(index=False))
```

```python
print("\n" + "="*70)
print("EXAMPLE QUERY 2: Financial Health Snapshot")
print("="*70)

result = conn.execute("""
    SELECT 
        ticker,
        MAX(CASE WHEN concept = 'Revenues' THEN value END) / 1000000000 as revenue_b,
        MAX(CASE WHEN concept = 'NetIncomeLoss' THEN value END) / 1000000000 as net_income_b,
        ROUND(
            MAX(CASE WHEN concept = 'NetIncomeLoss' THEN value END) * 100.0 
            / MAX(CASE WHEN concept = 'Revenues' THEN value END)
        , 1) as profit_margin_pct
    FROM sec_filings.financial_metrics
    WHERE concept IN ('Revenues', 'NetIncomeLoss')
    GROUP BY ticker
    ORDER BY ticker
""").df()

print(result.to_string(index=False))
```
