## Step 1: Load Data from PostgreSQL

```python
# Connect to PostgreSQL
conn = psycopg2.connect(
    host='localhost',
    port=5432,
    database='financial_data',
    user='postgres',
    password='postgres'
)
cursor = conn.cursor()

# Load financial metrics
cursor.execute('''
    SELECT ticker, concept, period_end, form, value 
    FROM sec_filings.financial_metrics
    ORDER BY ticker, period_end DESC
''')
metrics_rows = cursor.fetchall()
metrics_columns = ['ticker', 'concept', 'period_end', 'form', 'value']
metrics_data = [dict(zip(metrics_columns, row)) for row in metrics_rows]

print(f'✅ Loaded {len(metrics_data)} financial metrics')
```

```python
# Load MD&A chunks
cursor.execute('''
    SELECT id, ticker, year, section, text, text_length
    FROM sec_filings.filing_text_chunks
    ORDER BY ticker, year DESC
''')
chunks_rows = cursor.fetchall()
chunks_columns = ['id', 'ticker', 'year', 'section', 'text', 'text_length']
chunks_data = [dict(zip(chunks_columns, row)) for row in chunks_rows]

print(f'✅ Loaded {len(chunks_data)} MD&A chunks')
if chunks_data:
    print(f'   Subsections: {set(c["section"] for c in chunks_data)}')
else:
    print('   ⚠️  No chunks found')
```

---
## Step 2: Generate Embeddings for Chunks

```python
def get_embedding(text: str, model: str = "text-embedding-3-small") -> List[float]:
    """
    Get embedding vector for text using OpenAI API.
    text-embedding-3-small: 512 dimensions, fast & cheap
    """
    try:
        response = client.embeddings.create(
            input=text,
            model=model
        )
        return response.data[0].embedding
    except Exception as e:
        print(f"Error getting embedding: {e}")
        return None

print('✅ Defined embedding function')
print('   Note: This will call OpenAI API for each chunk')
print(f'   Cost: ~$0.02 per 1M tokens (~5K chunks)')
```

```python
if not chunks_data:
    print('❌ No chunks to embed')
else:
    print(f"\n🔄 Generating embeddings for {len(chunks_data)} chunks...")
    print("   (This may take 1-2 minutes)\n")
    
    embeddings = []
    for i, chunk in enumerate(chunks_data, 1):
        if i % 10 == 0:
            print(f"   [{i}/{len(chunks_data)}] Processing chunk {i}...")
        
        # Get embedding
        embedding = get_embedding(chunk['text'])
        if embedding:
            chunk['embedding'] = embedding
            embeddings.append(embedding)
        else:
            print(f"   ⚠️  Failed to embed chunk {i}")
    
    print(f"\n✅ Generated {len(embeddings)} embeddings")
    print(f"   Embedding dimension: {len(embeddings[0]) if embeddings else 0}")
    print(f"   Model: text-embedding-3-small")
```

---
## Step 3: Build Vector Search Index

```python
def cosine_similarity(vec1: List[float], vec2: List[float]) -> float:
    """Calculate cosine similarity between two vectors."""
    a = np.array(vec1)
    b = np.array(vec2)
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

def vector_search(query: str, top_k: int = 3, section_filter: Optional[str] = None) -> List[Dict]:
    """
    Semantic search using vector embeddings.
    
    Args:
        query: Natural language question
        top_k: Number of results to return
        section_filter: Filter by MD&A subsection
    
    Returns:
        List of most similar chunks with metadata
    """
    # Get query embedding
    query_embedding = get_embedding(query)
    if not query_embedding:
        return []
    
    # Calculate similarity to all chunks
    similarities = []
    for i, chunk in enumerate(chunks_data):
        if section_filter and chunk['section'] != section_filter:
            continue
        
        if 'embedding' not in chunk:
            continue
        
        similarity = cosine_similarity(query_embedding, chunk['embedding'])
        similarities.append((similarity, i, chunk))
    
    # Sort by similarity and return top_k
    similarities.sort(reverse=True, key=lambda x: x[0])
    
    retrieved = []
    for similarity, idx, chunk in similarities[:top_k]:
        retrieved.append({
            'ticker': chunk['ticker'],
            'year': chunk['year'],
            'section': chunk['section'],
            'text': chunk['text'][:500],
            'full_text': chunk['text'],
            'similarity_score': float(similarity)
        })
    
    return retrieved

print('✅ Defined vector search functions')
print('   Using cosine similarity for semantic matching')
```

---
## Step 4: Agentic RAG with Vector Search

```python
def get_metrics_context(ticker: str, metric: str) -> str:
    """Get metric values for a company."""
    matching = [m for m in metrics_data 
                if m['ticker'] == ticker.upper() and metric in m['concept']]
    
    if not matching:
        return f"No {metric} data found for {ticker}"
    
    lines = [f"\n{metric} for {ticker}:"]
    for m in sorted(matching, key=lambda x: x['period_end'], reverse=True)[:5]:
        value_b = m['value'] / 1000000000 if m['value'] > 1000000000 else m['value'] / 1000000
        unit = 'B' if m['value'] > 1000000000 else 'M'
        lines.append(f"  {m['period_end']} ({m['form']}): ${value_b:.2f}{unit}")
    
    return '\n'.join(lines)

class VectorRAGAgent:
    def __init__(self):
        self.conversation_history = []
        self.system_prompt = """You are a financial analysis expert with access to:
1. SEC filing data (10-K documents)
2. Financial metrics (revenue, assets, cash flow, etc.)
3. Management Discussion & Analysis (MD&A)

Your role is to answer questions about companies' financial performance, risks, and strategies.
Be specific, cite sources (ticker/year/section), and provide quantitative context when available.
If data is not available, clearly state that."""
    
    def ask(self, query: str) -> str:
        """Ask using vector-based semantic search."""
        # Route to relevant section
        section_filter = None
        if any(word in query.lower() for word in ['risk', 'threat', 'challenge']):
            section_filter = 'Risk Factors'
        elif any(word in query.lower() for word in ['revenue', 'sales', 'income', 'earnings']):
            section_filter = 'Results of Operations'
        elif any(word in query.lower() for word in ['cash', 'liquidity', 'capital']):
            section_filter = 'Liquidity & Capital Resources'
        elif any(word in query.lower() for word in ['balance', 'assets', 'liability']):
            section_filter = 'Financial Condition'
        
        # Semantic search using vectors
        retrieved_chunks = vector_search(query, top_k=3, section_filter=section_filter)
        
        # Build context
        context_parts = []
        if retrieved_chunks:
            context_parts.append("\nRelevant MD&A excerpts (ranked by semantic similarity):")
            for chunk in retrieved_chunks:
                context_parts.append(f"\n[{chunk['ticker']} {chunk['year']} - {chunk['section']} | Similarity: {chunk['similarity_score']:.2f}]")
                context_parts.append(chunk['text'])
        
        # Add metrics
        words = query.upper().split()
        tickers = [w for w in words if w in set(m['ticker'] for m in metrics_data)]
        
        for ticker in tickers[:3]:
            for metric in ['Revenues', 'NetIncomeLoss', 'Assets']:
                metric_context = get_metrics_context(ticker, metric)
                if 'No' not in metric_context:
                    context_parts.append(f"\n{metric_context}")
        
        context = '\n'.join(context_parts) if context_parts else "No relevant data found."
        
        user_message = f"""User Question: {query}

Available Context (from semantic search):
{context}

Please answer the user's question using the provided context. Be specific and cite sources."""
        
        self.conversation_history.append({
            "role": "user",
            "content": user_message
        })
        
        response = client.chat.completions.create(
            model="gpt-4-turbo",
            max_tokens=1024,
            messages=[
                {"role": "system", "content": self.system_prompt}
            ] + self.conversation_history
        )
        
        assistant_message = response.choices[0].message.content
        self.conversation_history.append({
            "role": "assistant",
            "content": assistant_message
        })
        
        return assistant_message

agent = VectorRAGAgent()
print('✅ Vector RAG Agent initialized')
```
