# Phase 2A: Extract & Load XBRL Financial Metrics to PostgreSQL

Fetch from SEC API and load into PostgreSQL.

**Prerequisites:**
- PostgreSQL running locally
- Database: `financial_data` created
- User: `postgres`

```python
import sys
sys.path.insert(0, '/Users/prajwalchambenandeeshappa/Github_Repos/Stocks_Earnings_Intelligence_Agent-Text2SQL/learning')

import os
import dlt
import logging
from typing import List, Dict
from datetime import datetime
from sec_api_orchestrator import FinancialDataPipeline

# Set PostgreSQL credentials as environment variables
os.environ['DESTINATION__POSTGRES__CREDENTIALS__HOST'] = 'localhost'
os.environ['DESTINATION__POSTGRES__CREDENTIALS__PORT'] = '5432'
os.environ['DESTINATION__POSTGRES__CREDENTIALS__DATABASE'] = 'financial_data'
os.environ['DESTINATION__POSTGRES__CREDENTIALS__USERNAME'] = 'postgres'
os.environ['DESTINATION__POSTGRES__CREDENTIALS__PASSWORD'] = 'postgres'

logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
print('✅ Imports successful')
```

```python
TICKERS = ['MSFT', 'AAPL', 'GOOGL']
KEY_METRICS = ['Revenues', 'NetIncomeLoss', 'OperatingIncomeLoss', 'Assets', 'Liabilities', 'StockholdersEquity']

print('🚀 PHASE 2A: Fetching from SEC API')
print(f'\n📊 Fetching {len(KEY_METRICS)} metrics for {len(TICKERS)} companies...\n')

all_metrics_data = []
all_filings_data = []

for ticker in TICKERS:
    print(f'📊 {ticker}...')
    pipeline_obj = FinancialDataPipeline(ticker)
    
    for metric in KEY_METRICS:
        metric_data = pipeline_obj.get_metric(metric, limit=10)
        all_metrics_data.extend(metric_data)
    
    filings_data = pipeline_obj.get_filings(limit=5)
    all_filings_data.extend(filings_data)

print(f'\n✅ Data fetched:')
print(f'   Metrics: {len(all_metrics_data)} records')
print(f'   Filings: {len(all_filings_data)} records')
```

```python
def transform_metric_for_dlt(records: List[Dict]) -> List[Dict]:
    transformed = []
    load_time = datetime.now().isoformat()
    for record in records:
        transformed.append({
            'ticker': record.get('ticker'),
            'concept': record.get('concept'),
            'period_end': record.get('period_end'),
            'filing_date': record.get('filing_date'),
            'form': record.get('form'),
            'value': int(record.get('value', 0)),
            'load_timestamp': load_time,
        })
    return transformed

def transform_filings_for_dlt(records: List[Dict]) -> List[Dict]:
    transformed = []
    load_time = datetime.now().isoformat()
    for record in records:
        transformed.append({
            'ticker': record.get('ticker'),
            'form': record.get('form'),
            'filing_date': record.get('filing_date'),
            'period_end': record.get('period_end'),
            'accession_number': record.get('accession_number'),
            'filing_url': record.get('filing_url'),
            'load_timestamp': load_time,
        })
    return transformed

metrics_transformed = transform_metric_for_dlt(all_metrics_data)
filings_transformed = transform_filings_for_dlt(all_filings_data)

print('✅ Data transformed')
```

```python
# Create dlt pipeline for PostgreSQL
pipeline = dlt.pipeline(
    pipeline_name='financial_data_pipeline',
    destination='postgres',
    dataset_name='sec_filings'
)

print('📝 Loading metrics to PostgreSQL...')
pipeline.run(
    metrics_transformed,
    table_name='financial_metrics',
    write_disposition='replace'
)
print(f'   ✅ {len(metrics_transformed)} records loaded')

print('\n📝 Loading filing metadata to PostgreSQL...')
pipeline.run(
    filings_transformed,
    table_name='sec_filings_metadata',
    write_disposition='replace'
)
print(f'   ✅ {len(filings_transformed)} records loaded')

print('\n' + '='*70)
print('✅ PHASE 2A COMPLETE!')
print('='*70)
print('\n📝 Next steps:')
print('   1. Close this notebook')
print('   2. Run: PHASE_2B_LOAD_FILING_TEXT.ipynb')
print('='*70)
```
