```python
#Filtering to a single ticker actually excludes the other companies' chunks
results = index.search(
    query="gross margin change",
    filter_dict={"ticker": "MSFT"},
    boost_dict={"text": 1.0},
    num_results=3,
)
for r in results:
    print(r["ticker"], "-", r["text"][:80])
```

```python
# ===== LESSON 5: PROMPT BUILDING =====

# 1. Define the prompt template
PROMPT_TEMPLATE = """You are a financial data expert. Answer the QUESTION using 
ONLY the information in CONTEXT below. If the context does not contain enough 
information to answer, say "I don't have enough information in the provided 
filings to answer that" — do not use outside knowledge or guess a number.

CONTEXT:
{context}

QUESTION: {question}

ANSWER:"""

# 2. Function to build prompts from search results
def build_prompt(question: str, search_results: list[dict]) -> str:
    """Join search results into a context block and format the prompt."""
    context = "\n\n".join(r["text"] for r in search_results)
    return PROMPT_TEMPLATE.format(context=context, question=question)

# 3. Reusable LLM function
def call_llm(prompt: str, model: str = "gpt-4o-mini", temperature: float = 0.0) -> str:
    """Call OpenAI API with the given prompt."""
    from openai import OpenAI
    
    client = OpenAI()
    response = client.chat.completions.create(
        model=model,
        temperature=temperature,
        messages=[{"role": "user", "content": prompt}],
    )
    return response.choices[0].message.content
```

```python
# ===== USAGE EXAMPLE =====
# showing how retrieval (minsearch) + generation (LLM) beats hallucination.

# Your question
question = "What was Microsoft's gross margin improvement in Q4 2024?"

# Search for relevant chunks
results = index.search(
    query=question,
    filter_dict={"ticker": "MSFT"},
    boost_dict={"text": 1.0},
    num_results=3,
)

# Build the prompt
prompt = build_prompt(question, results)

print("=== Generated Prompt ===")
print(prompt)
print("\n" + "="*60 + "\n")

# Call the LLM
answer = call_llm(prompt)
print("=== LLM Response ===")
print(answer)
```

```python
from rag_pipeline import RAGPipeline
from dotenv import load_dotenv

load_dotenv()
print("Dependencies loaded")
```

```python
# Create the RAG pipeline - it handles retrieval, prompt building, and generation
rag = RAGPipeline(
    documents=documents,
    llm_model="gpt-4o-mini",
    llm_temperature=0.0,
)

print("RAG Pipeline initialized")
```

```python
# Example 1: Query Microsoft's gross margin
question_1 = "What was Microsoft's gross margin improvement in Q4 2024?"

answer_1 = rag.answer(
    question=question_1,
    filter_dict={"ticker": "MSFT"},
    num_results=3,
)

print(f"Q: {question_1}")
print(f"A: {answer_1}")
```

```python
# Example 2: Get full context along with answer
question_2 = "What were Apple's revenue and margin trends?"

result = rag.answer_with_context(
    question=question_2,
    filter_dict={"ticker": "AAPL"},
    num_results=2,
)

print(f"Question: {result['question']}")
print(f"\nAnswer: {result['answer']}")
print(f"\nDocuments Retrieved: {len(result['context'])}")
for i, doc in enumerate(result['context'], 1):
    print(f"  {i}. {doc['ticker']} - {doc['fiscal_period']}")
```

```python
# Example 3: Query without ticker filter to get cross-company analysis
question_3 = "Which companies increased their gross margins?"

answer_3 = rag.answer(
    question=question_3,
    filter_dict=None,  # No ticker filter - searches all companies
    num_results=5,
)

print(f"Q: {question_3}")
print(f"A: {answer_3}")
```

```python
### Class based Benefits:
1. **Encapsulation**: Each class handles one responsibility (LLMClient, DocumentIndex, PromptBuilder)
2. **Reusability**: Create one pipeline, use for multiple questions
3. **Maintainability**: Changes to logic are centralized
4. **Testability**: Each class can be tested independently
5. **Extensibility**: Easy to add new features (e.g., caching, different LLMs, custom prompts)
```
